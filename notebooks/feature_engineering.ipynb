```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/cleaned_sales.csv")

df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Date Data Type:", df["Date"].dtype)
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek

df[["Date", "Year", "Month", "Day", "DayOfWeek"]].head()
```

```python
df["IsWeekend"] = df["DayOfWeek"].isin([5, 6]).astype(int)

df[["Date", "DayOfWeek", "IsWeekend"]].head(10)
```

```python
df["Revenue"] = df["Units Sold"] * df["Unit Price"]

df[["Units Sold", "Unit Price", "Revenue"]].head()
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Create Revenue feature
df["Revenue"] = df["Units Sold"] * df["Price"]

# Check result
df[["Units Sold", "Price", "Revenue"]].head()
```

```python
df["Discount Amount"] = df["Price"] * df["Discount"]

df[["Price", "Discount", "Discount Amount"]].head()
```

```python
df["Net Revenue"] = df["Revenue"] - df["Discount Amount"]

df[["Revenue", "Discount Amount", "Net Revenue"]].head()
```

```python
df = df.sort_values(
    by=["Store ID", "Product ID", "Date"]
).reset_index(drop=True)

df[["Store ID", "Product ID", "Date", "Units Sold"]].head(10)
```

```python
df["Lag_1_Day"] = (
    df.groupby(["Store ID", "Product ID"])["Units Sold"]
      .shift(1)
)

df[["Store ID", "Product ID", "Date", "Units Sold", "Lag_1_Day"]].head(10)
```

```python
df["Lag_7_Days"] = (
    df.groupby(["Store ID", "Product ID"])["Units Sold"]
      .shift(7)
)

df[[
    "Store ID",
    "Product ID",
    "Date",
    "Units Sold",
    "Lag_1_Day",
    "Lag_7_Days"
]].head(15)
```

```python
df["Rolling_7_Days"] = (
    df.groupby(["Store ID", "Product ID"])["Units Sold"]
      .transform(lambda x: x.shift(1).rolling(7).mean())
)

df[[
    "Store ID",
    "Product ID",
    "Date",
    "Units Sold",
    "Lag_1_Day",
    "Lag_7_Days",
    "Rolling_7_Days"
]].head(15)
```

```python
print("Missing Values in Feature Engineered Data:")
print(df.isnull().sum())
```

```python
df = df.dropna(
    subset=["Lag_1_Day", "Lag_7_Days", "Rolling_7_Days"]
).reset_index(drop=True)

print("Shape after removing lag missing values:", df.shape)
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
print("Dataset Shape:", df.shape)

print("\nFeature Columns:")
print(df.columns.tolist())

print("\nSample Data:")
df.head()
```

```python
df.to_csv("../data/feature_engineered_data.csv", index=False)

print("Feature-engineered dataset saved successfully!")
```
